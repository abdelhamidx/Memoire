# Suite de tests automatisés — PFE

Ce notebook exécute la suite `pytest` du projet (`tests/test_api.py`) et
affiche le résultat complet, pour disposer d'une preuve horodatée et
lisible de la couverture de tests lors de la soutenance.

La suite tape la vraie base de données (pas de mocks) : elle couvre
- les endpoints publics (`/health`, `/stations`, `/stations/{id}`, `/stations/{id}/history`)
- les prédictions, legacy et multi-horizon (`/predict/{id}`, `/predictions`, validation de `horizon_minutes`)
- le rééquilibrage (`/rebalancing`)
- l'authentification et les rôles (`/auth/login`, `/auth/me`, `/admin/users`, contrôle d'accès admin vs chauffeur)


```python
import subprocess
import sys
import os

project_root = os.path.abspath("..")

result = subprocess.run(
    [sys.executable, "-m", "pytest", "tests/", "-v", "--tb=short", "--color=no"],
    cwd=project_root,
    capture_output=True,
    text=True,
)

print(result.stdout)
if result.stderr:
    print("--- stderr ---")
    print(result.stderr)
print(f"Code de sortie : {result.returncode}")
```

## Pour le mémoire

- Suite pytest tapant la vraie base (pas de mocks) : montre que le pipeline
  bout en bout (auth, stations, prédictions multi-horizon, rééquilibrage)
  fonctionne réellement, pas seulement en théorie.
- Couvre explicitement les cas d'erreur (404, 401, 403, 409, 422), pas
  seulement les cas heureux — important pour montrer une démarche de
  test sérieuse en soutenance.
- Cette même commande (`pytest tests/`) est celle exécutée automatiquement
  par le pipeline CI/CD (`.github/workflows/ci.yml`) à chaque push.
